## Overview
1) **Continuous trading vs. Buy-And-Hold strategy**\
    Use either a provided Python notebook or Excel sheet to simulate the wealth process in the log-utility problem, i.e. use the constant portfolio process
    $$\pi_t = \frac{b - r}{\sigma^2}$$
    where $r$ is the riskless interest rate and $b$, $\sigma$ are the drift and the volatility of the stock price. Compare this to the wealth process of a Buy-And-Hold strategy that starts by assigning
    $$x\frac{b - r}{\sigma^2}$$
    in the stock and
    $$x \left(1 - \frac{b - r}{\sigma^2}\right)$$
    in the money market account
    at time $t=0$, and do not change the stock and bond position until time $T$. Compare the differences (for the same price paths!) for the varying maturities $T \in \{0.1, 1, 5\}$.

2) **CPPI-Strategy**\
    Use either a provided Python notebook or Excel sheet to illustrate the wealth process of a CPPI-strategy and examine the possible GAP-Risk by choosing extreme values for the multiplier $M$ and possibly a rough time discretization. Demonstrate also that the GAP-Risk nearly disappears for a fine discretization.

3) **Log-optimal Strategy vs. Value Preserving Strategies (Bonus)**\
    Implement the optimal solution of the log-utility portfolio and consumption strategy and of the value preserving strategy. Compare the simulations for varying choices of the market coefficients and the maturity time $T$.
